## Q3

```python
# 3a
letters = ['a', 'b', 'c', 'd']

# 3b: '3' * 2 gives '33'. int('33') // 11 gives 3.
# The outer int() keeps 3, so index 3 gives the fourth item, 'd'.
print("3b:", letters[int(int('3' * 2) // 11)])

# 3c: -1 means the last item, so the result is 'd'.
print("3c:", letters[-1])

# 3d: [:2] takes indexes 0 and 1, so the result is ['a', 'b'].
print("3d:", letters[:2])
```

## Q5



```python
def comma_code(items):
    if len(items) == 0:
        return ''
    elif len(items) == 1:
        return str(items[0])
    elif len(items) == 2:
        return str(items[0]) + ' and ' + str(items[1])

    result = ''
    for i in range(len(items) - 1):
        result = result + str(items[i]) + ', '
    result = result + 'and ' + str(items[-1])
    return result

spam = ['apples', 'bananas', 'tofu', 'cats']
print(comma_code(spam))
```
