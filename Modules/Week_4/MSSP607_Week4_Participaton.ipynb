### 1.a

```python
import pandas as pd

import pandas as pd

df = pd.read_csv("/parsed.csv")
df[["mag", "magType", "parsed_place", "tsunami"]].head()
```

### 2.a


```python
japan_mb = df[(df["parsed_place"] == "Japan") & (df["magType"] == "mb")]
percentile_95 = japan_mb["mag"].quantile(0.95)
print("95th percentile:", percentile_95)
```

The 95th percentile of magnitude in Japan for `mb` is **4.9**.


### 2.b


```python
indonesia = df[df["parsed_place"] == "Indonesia"]
tsunami_count = indonesia["tsunami"].sum()
tsunami_percentage = tsunami_count / len(indonesia) * 100
print("Tsunami flags:", tsunami_count)
print("Total Indonesia records:", len(indonesia))
print(f"Percentage: {tsunami_percentage:.2f}%")
```

**34 out of 147** records in Indonesia have a tsunami flag. The percentage is **23.13%**.


### 2.c


```python
nevada = df[df["parsed_place"] == "Nevada"]
nevada.describe()
```

Nevada has **681** records. The mean magnitude is about **0.50**, and the median is **0.40**. The minimum is **-0.50**, and the maximum is **2.90**.


### 2.d


```python
ring_locations = [
    "Bolivia", "Chile", "Ecuador", "Peru", "Costa Rica", "Guatemala",
    "Mexico", "Japan", "Philippines", "Indonesia", "New Zealand",
    "Canada", "Fiji", "Alaska", "Washington", "California", "Russia",
    "Taiwan", "Tonga", "Kermadec Islands"
]

df["ring_of_fire"] = (
    df["parsed_place"].isin(ring_locations)
    | df["parsed_place"].str.contains("Antarctic", na=False)
)
df[["parsed_place", "ring_of_fire"]].head()
```

I added `ring_of_fire` to mark the listed locations. Exact matching excludes New Mexico, and the Antarctic search includes the Antarctic locations.


### 2.e


```python
inside_count = df["ring_of_fire"].sum()
outside_count = (~df["ring_of_fire"]).sum()
print("Inside Ring of Fire:", inside_count)
print("Outside Ring of Fire:", outside_count)
```

There are **7,188** records inside the listed Ring of Fire locations and **2,144** outside them.


### 2.f


```python
ring_tsunami_count = df.loc[df["ring_of_fire"], "tsunami"].sum()
print("Tsunami flags along the Ring of Fire:", ring_tsunami_count)
```

There are **45** tsunami flags in the Ring of Fire locations.
